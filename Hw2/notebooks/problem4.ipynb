Using a second order Taylor expansion of the stage cost $c(\tilde{s}_k, \tilde{u}_k)$ around $(\bar{s}_k, \bar{u}_k)$:

$$
c(\tilde{s}_k, \tilde{u}_k) = c(\bar{s}_k, \bar{u}_k)
+ \left(\underbrace{\frac{\partial c}{\partial s}\bigg|_{(\bar{s}_k,\bar{u}_k)}}_{=\,q_k}\right)^\top\tilde{s}_k
+ \left(\underbrace{\frac{\partial c}{\partial u}\bigg|_{(\bar{s}_k,\bar{u}_k)}}_{=\,r_k}\right)^\top\tilde{u}_k
+ \frac{1}{2}\tilde{s}_k^\top\underbrace{\frac{\partial^2 c}{\partial s^2}\bigg|_{(\bar{s}_k,\bar{u}_k)}}_{=\,Q}\tilde{s}_k
+ \frac{1}{2}\tilde{u}_k^\top\underbrace{\frac{\partial^2 c}{\partial u^2}\bigg|_{(\bar{s}_k,\bar{u}_k)}}_{=\,R}\tilde{u}_k
$$

hence $\boxed{q_k = Q(\bar{s}_k - s_\text{goal})}$ and $\boxed{r_k = R\bar{u}_k}$.

Same for the terminal cost $c_N(\tilde{s}_N)$:

$$
c_N(\tilde{s}_N) = c_N(\bar{s}_N) + \left(\underbrace{\frac{\partial c_N}{\partial s}\bigg|_{\bar{s}_N}}_{=\,q_N}\right)^\top\tilde{s}_N + \frac{1}{2}\tilde{s}_N^\top\underbrace{\frac{\partial^2 c_N}{\partial s^2}\bigg|_{\bar{s}_N}}_{=\,Q_N}\tilde{s}_N
$$

hence $\boxed{q_N = Q_N(\bar{s}_N - s_\text{goal})}$.
