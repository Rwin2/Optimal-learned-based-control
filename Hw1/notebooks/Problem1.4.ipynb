## Problem 1.4 — Zermelo's ship

### a)

Minimum-time objective: $J = \int_{t_0}^{t_1} 1\, dt = t_1 - t_0$, so the Hamiltonian is

$$H = 1 + p_x\!\left(v\cos\theta(t) + \frac{v}{h}y(t)\right) + p_y\, v\sin\theta(t)$$

**State equations** (dynamic constraints):

$$\dot{x}^*(t) = \begin{bmatrix} \dot{x} \\ \dot{y} \end{bmatrix} = \begin{bmatrix} v\cos\theta^* + \dfrac{v}{h}y \\ v\sin\theta^* \end{bmatrix}$$

**Costate (adjoint) equations**:

$$\dot{p}^*(t) = \begin{bmatrix} -\dfrac{\partial H}{\partial x} \\[6pt] -\dfrac{\partial H}{\partial y} \end{bmatrix} = \begin{bmatrix} 0 \\[4pt] -p_x\dfrac{v}{h} \end{bmatrix}$$

Since $p_x$ is constant:

$$p_x = \text{const}, \qquad p_y(t) = -p_x\frac{v}{h}\,t + \beta \quad \text{for some } \beta \in \mathbb{R}$$

**Optimality condition** ($\theta$ unconstrained):

$$\frac{\partial H}{\partial \theta} = 0 \quad\Longrightarrow\quad 0 = -p_x\, v\sin\theta^*(t) + p_y\, v\cos\theta^*(t)$$

$$\Longrightarrow\quad \tan\theta^*(t) = \frac{p_y(t)}{p_x} = \frac{-p_x\frac{v}{h}\,t + \beta}{p_x} = -\frac{v}{h}\,t + \frac{\beta}{p_x}$$

Setting $\alpha := \frac{\beta}{p_x} \in \mathbb{R}$:

$$\boxed{\;\tan\theta^*(t) = \alpha - \frac{v}{h}\,t\;}$$

**Transversality condition** (free terminal time):

Substituting $p_y(t_1^*) = -p_x\dfrac{v}{h}t_1^* + \beta$ (from the costate equation) into $H$, and using $y(t_1^*) = 0$:

$$
\begin{aligned}
H^*(t_1^*) &= 1 + p_x\!\left(v\cos\theta^*(t_1^*) + \frac{v}{h}\underbrace{y(t_1^*)}_{=\,0}\right) + \left(-p_x\frac{v}{h}t_1^* + \beta\right)v\sin\theta^*(t_1^*) \\
&= 1 + v\Bigl[p_x\cos\theta^*(t_1^*) + \underbrace{\left(-p_x\frac{v}{h}t_1^* + \beta\right)}_{=\,p_x\tan\theta^*(t_1^*)}\sin\theta^*(t_1^*)\Bigr] \\
&= 1 + vp_x\underbrace{\left(\cos\theta^* + \frac{\sin^2\theta^*}{\cos\theta^*}\right)}_{=\,1/\cos\theta^*} \\
&= 1 + \frac{v\,p_x}{\cos\theta^*(t_1^*)} = 0
\end{aligned}
$$

$$\Longrightarrow\quad p_x = -\frac{\cos\theta^*(t_1^*)}{v}$$

### b)

With $w(y(t)) \equiv \beta$ constant, the Hamiltonian becomes $H = 1 + p_x(v\cos\theta + \beta) + p_y v\sin\theta$, so

$$\dot{p}^*(t) = \begin{bmatrix} -\dfrac{\partial H}{\partial x} \\[4pt] -\dfrac{\partial H}{\partial y} \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix} \quad\Longrightarrow\quad p_x^* \in \mathbb{R},\; p_y^* \in \mathbb{R} \text{ (both constant)}$$

Optimality condition:

$$\frac{\partial H}{\partial \theta} = 0 \quad\Longrightarrow\quad \tan\theta^*(t) = \frac{p_y^*}{p_x^*} = \text{const} \quad\Longrightarrow\quad \theta^* = \text{const} := \theta_0$$

**Transversality condition** (free terminal time):

From the optimality condition $p_y = p_x\tan\theta_0$, substituting into $H$:

$$
\begin{aligned}
H^*(t_1^*) &= 1 + p_x(v\cos\theta_0 + \beta) + p_x\tan\theta_0\, v\sin\theta_0 \\
&= 1 + p_x\left(v\cos\theta_0 + \beta + \frac{v\sin^2\theta_0}{\cos\theta_0}\right) \\
&= 1 + p_x\cdot\frac{v + \beta\cos\theta_0}{\cos\theta_0} = 0
\end{aligned}
$$

$$\Longrightarrow\quad p_x = -\frac{\cos\theta_0}{v + \beta\cos\theta_0}$$

Since $\theta_0$ is constant, the trajectory integrates directly:

$$x(t) = \bigl(v\cos\theta_0 + \beta\bigr)(t - t_0) + x_0, \qquad y(t) = v\sin\theta_0\,(t - t_0) + y_0$$

Apply terminal BCs $x(t_1^*) = 0$, $y(t_1^*) = 0$. Let $T := t_1^* - t_0$:

$$\bigl(v\cos\theta_0 + \beta\bigr)T = -x_0 \qquad (1)$$
$$v\sin\theta_0\; T = -y_0 \qquad (2)$$

Squaring and adding $(1)^2 + (2)^2$: $(vT)^2 = (x_0 + \beta T)^2 + y_0^2$, which expands to

$$\boxed{\;(v^2 - \beta^2)\,T^2 - 2\beta x_0\,T - (x_0^2 + y_0^2) = 0\;}$$

Solving by the quadratic formula $T = \dfrac{2\beta x_0 \pm \sqrt{4\bigl[v^2 x_0^2 + (v^2 - \beta^2)y_0^2\bigr]}}{2(v^2-\beta^2)}$:

---

**Case (i) $v > \beta$:** leading coefficient $> 0$, discriminant $= 4\bigl[v^2 x_0^2 + (v^2-\beta^2)y_0^2\bigr] > 0$ always. Taking the positive root:

$$\boxed{\;t_1^* - t_0 = \frac{\beta x_0 + \sqrt{v^2 x_0^2 + (v^2 - \beta^2)\,y_0^2}}{v^2 - \beta^2}\;}$$

No restriction on $x_0$.

---

**Case (ii) $v = \beta$:** the $T^2$ term vanishes, leaving a linear equation $-2\beta x_0\,T - (x_0^2+y_0^2) = 0$:

$$\boxed{\;t_1^* - t_0 = -\frac{x_0^2 + y_0^2}{2\beta\, x_0}\;}$$

Requires $x_0 < 0$ for $T > 0$ (ship must start upstream of origin — it cannot make negative headway against the equal current).

---

**Case (iii) $v < \beta$:** leading coefficient $< 0$; the discriminant requires $v^2 x_0^2 \ge (\beta^2 - v^2)\,y_0^2$ for real solutions. Taking the smaller positive root:

$$\boxed{\;t_1^* - t_0 = \frac{\beta x_0 + \sqrt{v^2 x_0^2 - (\beta^2 - v^2)\,y_0^2}}{v^2 - \beta^2}\;}$$

Requires $x_0 < 0$ and $v^2 x_0^2 \ge (\beta^2 - v^2)\,y_0^2$ (the current is too strong to overcome; ship must start sufficiently upstream).
