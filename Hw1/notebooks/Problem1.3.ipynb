## Problem 1.3 — Extremal curves

### 1.3

Using the Euler-Lagrange equation in the Fundamental Theorem of Calculus of Variations as a necessary condition for an optimal curve:

$$\frac{\partial}{\partial x}\left(\frac{1}{2}\dot{x}(t)^2 + 5x(t)\dot{x}(t) + x(t)^2 + 5x(t)\right) - \frac{d}{dt}\frac{\partial}{\partial \dot{x}}\left(\frac{1}{2}\dot{x}(t)^2 + 5x(t)\dot{x}(t) + x(t)^2 + 5x(t)\right) = 0$$

$$5\dot{x}(t) + 2x(t) + 5 - \frac{d}{dt}\left[\dot{x}(t) + 5x(t)\right] = 0$$

$$5\dot{x}(t) + 2x(t) + 5 - \ddot{x}(t) - 5\dot{x}(t) = 0$$

$$\boxed{\;\ddot{x}(t) - 2x(t) = 5\;}$$

General solution for $A, B \in \mathbb{R}$:

$$x^*(t) = A\cosh(\sqrt{2}\,t) + B\sinh(\sqrt{2}\,t) - \frac{5}{2}$$

with BC $x^*(0) = 1$ and $x^*(1) = 3$.

$$x^*(0) = A - \frac{5}{2} = 1 \quad\Longrightarrow\quad A = \frac{7}{2}$$

$$x^*(1) = \frac{7}{2}\cosh(\sqrt{2}) + B\sinh(\sqrt{2}) - \frac{5}{2} = 3 \quad\Longrightarrow\quad B = \frac{\dfrac{11}{2} - \dfrac{7}{2}\cosh(\sqrt{2})}{\sinh(\sqrt{2})}$$

$$\boxed{\;x^*(t) = \frac{7}{2}\cosh(\sqrt{2}\,t) + \frac{11 - 7\cosh(\sqrt{2})}{2\sinh(\sqrt{2})}\sinh(\sqrt{2}\,t) - \frac{5}{2}\;}$$

Please see plot.

```python
import numpy as np
import matplotlib.pyplot as plt

A = 7 / 2
B = (11/2 - (7/2) * np.cosh(np.sqrt(2))) / np.sinh(np.sqrt(2))

def x_star(t):
    return A * np.cosh(np.sqrt(2) * t) + B * np.sinh(np.sqrt(2) * t) - 5/2

t = np.linspace(0, 1, 500)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(t, x_star(t), lw=2)
ax.scatter([0, 1], [1, 3], color='k', zorder=5)
ax.set_xlabel(r'$t$')
ax.set_ylabel(r'$x^*(t)$')
ax.set_title(r'Extremal curve for $J(x)$')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
